# Exposure Time Calculator

During observations of the selected target, the CCD camera records in addition to the object
photons, other sources that generate an electrical signal in the system. All electrical signals not
generated by photons from the source are characterized as noise. Specific sources of noise can
be defined as follows:
- Thermal noise (dark current): caused by the thermal motion of electrons in the chip.
- Read-out noise: a result of the electronics during the readout process of the image.
- Background noise: referred to also as sky background caused by the light pollution.
- Cosmic rays: detection of cosmic rays that create false detections.
- Pixel defects: incorrect pixel readings (bad semiconductor junctions).
The quality of the image is defined by a parameter known as the signal-to-noise ratio (SNR).
Higher SNR values indicate better image quality. An unwritten rule is to use data with an SNR
greater than 2 or 3. This raises the question: with a given setup (telescope + CCD), what is the
required exposure time to achieve a given SNR value? The answer is not straightforward, and
there is no absolute formula to precisely determine the exposure time. Instead, using certain
assumptions, one can make an estimate.

**Signal Noise:** In the simplest case, even if no other noise is assumed, some measurement error
(signal noise) still exists. This noise originates from the counting of photons arriving from the
source. Photon detection on the CCD chip is a random process that follows a Poisson
distribution, which gives that the registered noise is proportional to the square root of the
number of photons that reach the CCD chip during the observed period, i.e., exposure time.
With a linear increase in exposure time, the SNR grows with the square root of the exposure
time. The detected signal is described by the number of electrons per second - [Ssig] = e-/s .

**Thermal Noise:** It is well known that particles move at a given temperature, and electrons in
the semiconductor of the CCD chip behave similarly. Thermal noise or so-called dark current
is usually reduced by lowering the CCD's temperature. The lower the temperature, the smaller
the thermal noise. However, here we assume that the camera is at a certain temperature for
which the thermal noise must be considered. At this temperature, there is a finite probability
that an electron from the semiconductor will fall into a potential well. Thermal noise can be
removed during the calibration of images. In general, for non-calibrated images, the thermal
noise is a characteristic of the camera at a given temperature and is described by the number of
electrons generated in one hour per pixel - [Sdc] = e-/h/pix.



**Background Noise:** Light pollution is the primary source of the limiting magnitude of the
optical instrument being used, whether it is the human eye or a telescope. Additionally, the
Moon is another dominant source of background noise, which is why observational cycles are
planned during moonless nights or directed at parts of the sky more than 90° away from the
Moon. Background noise is a measurable quantity, and for a given location (depending on the
Moon's phase), it is given as the number of electrons per unit time that fall on one pixel - [Ssky]
= e-/s/pix. It is often expressed in magnitudes per square arcsecond - [Bsky] = mag/arcsec2
.

**Read-Out Noise:** During the readout process, the electronics can introduce additional electrons.
The number of generated electrons can be predicted and is known for each camera, and it does
not depend on the exposure time. The value of read-out noise is given as the number of
electrons generated per pixel during a single readout $-[Sro2] = e-/pix$.

In addition to the listed sources of noise, the SNR value is also affected by the so-called binning
of the CCD image, that involves combining the charge from multiple adjacent pixels on a CCD
into a single pixel before the image is read out. Binning changes the image size and resolution
but results in an increase in SNR (i.e., when reading a binned image, we register a larger
number of photons (electrons) for the same value of the read-out noise). Furthermore, the pixel
size of the CCD chip [u] = μm influences the image resolution. Naturally, the quantum
efficiencyof the CCD chip also plays a significant role as it it represents the ratio of generated
electrons to the number of photons that reach the CCD chip and is typically denoted as q.

Based on the above, we can derive an equation that gives the SNR for a specific observed
object. All sources of noise affecting the observation propagate quadratically, leading to the
following expression:

$$
\mathrm{SNR} =
\frac{S_{\mathrm{sig}}\,t}
{\sqrt{S_{\mathrm{sig}}\,t + R_{\mathrm{sky}}\,t + S_{\mathrm{dc}}\,t\,n + S_{\mathrm{ro}}^{2}\,n}}
$$

```python
# Calculate the pixel scale and the effective number of pixels in the source aperture.
import numpy as np
import matplotlib.pyplot as plt  

def get_pixel_scale(D, f_num, u):
    focal_length = f_num * D  # Telescope focal length [m]
    pixel_scale = 206265 * (u * 1e-6) / focal_length  # Angular size of one pixel [arcsec/pixel]
    return pixel_scale

def get_n_pixels(r, pixel_scale):
    n = np.pi * (r / pixel_scale)**2  # Effective number of pixels inside the circular aperture
    return n
```

```python

def get_source_rate(D, q, V):
    f0 = 1000  # Zero-magnitude flux [e-/s/cm^2/Å]
    delta_lambda = 900  # V-filter bandwidth [Å]
    area_cm2 = np.pi * (D * 100 / 2)**2  # Telescope collecting area [cm^2]

    source_flux = f0 * 10**(-0.4 * V) * delta_lambda  # Source flux across the filter band
    S_sig = source_flux * area_cm2 * q  # Detected source rate [e-/s]

    return S_sig
```

```python
# Calculate the sky background rate per pixel.
def get_sky_rate_per_pixel(D, f_num, u, q, B_sky):
    f0 = 1000  # Zero-magnitude flux [e-/s/cm^2/Å]
    delta_lambda = 900  # V-filter bandwidth [Å]

    area_cm2 = np.pi * (D * 100 / 2)**2  # Telescope collecting area [cm^2]
    pixel_scale = get_pixel_scale(D, f_num, u)
    pixel_area = pixel_scale**2  # Pixel area on the sky [arcsec^2]

    sky_flux = f0 * 10**(-0.4 * B_sky) * delta_lambda
    S_sky = sky_flux * area_cm2 * q * pixel_area  # Sky rate per pixel [e-/s/pixel]

    return S_sky
```

```python
# Calculate the exposure time required to reach the given SNR.
def get_exposure_time(S_sig, S_sky, S_dc, S_ro_sq, n, SNR):
    R_sky = S_sky * n  # Total sky rate within the aperture [e-/s]

    rate_sum = S_sig + R_sky + S_dc * n
    read_variance = S_ro_sq * n

    t = (
        SNR**2 * rate_sum
        + np.sqrt(SNR**4 * rate_sum**2 + 4 * S_sig**2 * SNR**2 * read_variance)
    ) / (2 * S_sig**2)

    return t  # Exposure time [s]
```

```python
# Calculate the exposure time for a point source from the input parameters.
def get_point_source_exposure_time(D, f_num, S_ro_sq, u, q, S_dc, r, SNR, V, B_sky):
    pixel_scale = get_pixel_scale(D, f_num, u)
    n = get_n_pixels(r, pixel_scale)

    S_sig = get_source_rate(D, q, V)
    S_sky = get_sky_rate_per_pixel(D, f_num, u, q, B_sky)

    t = get_exposure_time(S_sig, S_sky, S_dc, S_ro_sq, n, SNR)
    return t  # Exposure time [s]
```

```python
# Default parameters
D = 8
f_num = 2.77
S_ro_sq = 5
u = 27
q = 0.7
S_dc = 3
V = 25
B_sky = 22
r_default = 2
SNR_default = 10
```

```python
# Plot exposure time versus SNR for different aperture radii
snr_values = np.arange(3, 1004, 1)
r_values = [0.01, 0.1, 0.5, 1, 3]

plt.figure(figsize=(10, 6))

for r_value in r_values:
    t_values = []

    for snr in snr_values:
        t = get_point_source_exposure_time(
            D, f_num, S_ro_sq, u, q, S_dc, r_value, snr, V, B_sky
        )
        t_values.append(t)

    plt.plot(snr_values, t_values, label=f"r = {r_value} arcsec")

plt.xlabel("SNR")
plt.ylabel("Exposure time [s]")
plt.title("Exposure time vs. SNR for different aperture radii")
plt.legend()
plt.grid()
plt.show()
```

```python
# Plot exposure time versus magnitude for different telescope diameters
magnitude_values = np.arange(10, 30.5, 0.5)
diameter_values = [1, 2.5, 6, 11]

plt.figure(figsize=(10, 6))

for D_value in diameter_values:
    t_values = []

    for V_value in magnitude_values:
        t = get_point_source_exposure_time(
            D_value, f_num, S_ro_sq, u, q, S_dc,
            r_default, SNR_default, V_value, B_sky
        )
        t_values.append(t)

    plt.plot(magnitude_values, t_values, label=f"D = {D_value} m")

plt.xlabel("V magnitude")
plt.ylabel("Exposure time [s]")
plt.title("Exposure time vs. magnitude for different telescope diameters")
plt.legend()
plt.grid()
plt.show()
```

```python
# Plot exposure time versus sky brightness for different aperture radii
B_sky_values = np.linspace(18, 24, 61)
r_values = [0.01, 0.1, 0.5, 1, 3]

plt.figure(figsize=(10, 6))

for r_value in r_values:
    t_values = []

    for B_sky_value in B_sky_values:
        t = get_point_source_exposure_time(
            D, f_num, S_ro_sq, u, q, S_dc,
            r_value, SNR_default, V, B_sky_value
        )
        t_values.append(t)

    plt.plot(B_sky_values, t_values, label=f"r = {r_value} arcsec")

plt.xlabel(r"$B_{sky}$ [mag/arcsec$^2$]")
plt.ylabel("Exposure time [s]")
plt.title("Exposure time vs. sky brightness for different aperture radii")
plt.legend()
plt.grid()
plt.show()
```
