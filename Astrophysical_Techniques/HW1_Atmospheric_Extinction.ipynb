# Estimatin the Effect of Atmospheric Extinction

The emission coming from the astronomical sources is affected by the matter present between the source and observer, mainly by the interstellar matter (dust and gas) and by the Earth's atmosphere. To minimize the effects of the atmosphere, professional observatories are built on high and dry places with lower precipitation and less cloudy nights. Another, more challenging and expensive, solution to this problem is to launch telescopes in space, outside the influence of the Earth's atmosphere. The radiation of an astronomical object (e.g., a star) passing through the atmosphere is affected by refraction (which causes changes in the apparent position of a star) and atmospheric extinction, which attenuate the flux reaching the Earth's surface (denoted as e.g., f Al, while the flux at the top of the atmosphere is denoted with fl); here we neglect the absorption by photometric filters). 

The atmosphere has less impact on objects closer to the zenith than on those near the horizon because the light passes through a shorter geometric path, encountering fewer atmospheric layers, and thus suffers less extinction. The parameter that describes the thickness of the atmosphere is defined as **airmass X**. A higher airmass indicates that the object is closer to the horizon. The airmass is normalized to unity at the zenith X = 1. This is an imaginary boundary above which we consider having a vacuum, with no atmospheric effect. **As the zenith distance increases, the airmass grows**. Several models (for both plane-parallel and spherical atmospheres, see Mathar 2015) describe how airmass changes with the zenith distance of the object. In the first-order approximation for a spherical atmosphere, the expression is given by:

$$X = \frac{1}{\cos Z}\left(1 - 0.0012\left(\frac{1}{\cos^2 Z} - 1\right)\right)$$

where Z is the zenith distance of the object. This expression is better than the plane-parallel case because it can be used for larger zenith distances (the plane-parallel case is limited to Z = 70° or X = 3).

The Beer–Lambert law describes how light intensity decreases as it travels through an absorbing medium. It is given by, 

$$
I=I_0e^{-\alpha L}
$$

where $I_0$ is the initial Intensity, $I$ the observed intensity, $\alpha$ the attenuation coefficient, and $L$ the distance traveled through the atmosphere. The equation is telling us that the londer the distance L, the more light is absorbed or scattered, and the lower the observed intensity I. 

From our diagram we will rename the variables. H is the vertical thickness of the atmosphere, Z is the zenith distance, and L is the distance traveled by light. 

For a plane-parallel atmoshpere, we assume horizontal atmospheric layers with a vertical thickness. 

$$
\cos Z=\frac{H}{L}
$$

Therefore:

$$
L=\frac{H}{\cos Z}
$$

So as the zenith distance increases, the light travels through a longer atmospheric path. 

Now, substituting into the Beer–Lambert law we get that,

$$
I=I_0e^{-\alpha H/\cos Z}
$$

which describes how intensity changes depending on the zenith distance. 

At the zenith, $Z=0$, so:

$$
I_{\mathrm{zenith}}=I_0e^{-\alpha H}
$$

whcih represents the intensity after the light has traveled vertically through the atmosphere. 

The airmass is defined as the ratio between the atmospheric path length and the vertical path:

$$
X=\frac{L}{H}
$$

Substituting $L$:

$$
X=\frac{H/\cos Z}{H}
$$

Finally:

$$
X=\frac{1}{\cos Z}=\sec Z
$$

This shows that the airmass increases with zenith distance, meaning that light travels through more atmosphere and experiences greater extinction.

The **extinction** in the Atmosphere is caused by several mechanisms:
- **Rayleigh scattering on molecules:** this scattering is proportional to lambda-4
, and is responsible for the blue color of the sky. It changes minimally over time.
- **Ozone absorption:** does not change drastically over short time intervals; defines the short-wavelength limit of atmospheric transparency.
- **Aerosol scattering (Mie scattering):** proportional to lambda-1, and is responsible for the white-grey color of clouds. It varies significantly over time and has the greatest influence up to altitudes of about 4.5 km.
- **Molecular absorption:** the main absorbers are water vapor, carbon dioxide, and oxygen. Thesevary considerably with time, location, and temperature.Each of these processes depends on the wavelength of the radiation. The magnitude observed at the Earth’s surface can be described using the e**xtinction (Beer-Lambert) law**:

$$
m_{\lambda}^{A} = m_{\lambda} + k_{\lambda} X
$$

where $k_{\lambda}$ is the extinction coefficient for a given wavelength (or filter) and is assumed to remain constant over time and isotropic. Under these assumptions, we consider the atmosphere to be photometric. Based on this, professional observatories characterize if an observing night had photometric conditions, indicating most optimal observing conditions in terms of the atmosphere properties.

By observing a standard (or non-variable) star (or multiple stars) over the course of a single night at different airmasses, we can estimate the extinction coefficient and the magnitude outside the atmosphere using the above equation. In addition to the extinction coefficient that we can use to correct the magnitude, we can also obtain a correction coefficient for a given color index CI. However, this coefficient changes from star to star due to different spectrum slopes, requiring the introduction of a second-order extinction coefficient. The **extinction coefficient for the color index** is given by:

$$
k_{\lambda_1-\lambda_2} = k’ + k’’ CI
$$

where CI is the color index, and $k’$ and $k’’$ are the first- and second-order extinction coefficients, respectively.
