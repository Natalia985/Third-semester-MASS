## Quasar Outflow Velocities

We visually estimate the C IV absorption minima from the rest-frame spectra. Using a reference wavelength of 1549 Å, the approximate outflow speeds are:

$
v \approx c\,\frac{1549-\lambda_{\min}}{1549}.
$

These measurements describe the velocity at the selected absorption minimum, rather than the maximum outflow velocity.

The order will be the column first from the left and then second the one on the right. 

```python
c = 300000      
lambda_0 = 1549 # C IV reference wavelength, Å
```

```python
# SDSS J014552.59-083536.8
# No clear absorption trough: velocity cannot be measured.

print("J014552: no clear absorption minimum.")
```

```python
# SDSS J003832.26+152515.5
# Multiple absorption minima; approximate selected minimum.

lambda_min_2 = 1495
v_2 = c * (lambda_0 - lambda_min_2) / lambda_0

print(f"J003832: ≈ {v_2:.0f} km/s")
```

```python
# SDSS J005010.54+153909.5

lambda_min_3 = 1530
v_3 = c * (lambda_0 - lambda_min_3) / lambda_0

print(f"J005010: ≈ {v_3:.0f} km/s")
```

```python
# SDSS J015024.43+004433.0

lambda_min_4 = 1530
v_4 = c * (lambda_0 - lambda_min_4) / lambda_0

print(f"J015024: ≈ {v_4:.0f} km/s")
```

```python
# SDSS J075757.92+245128.8

lambda_min_5 = 1495
v_5 = c * (lambda_0 - lambda_min_5) / lambda_0

print(f"J075757: ≈ {v_5:.0f} km/s")
```

```python
# SDSS J102038.73+094259.6
# Narrow absorption feature; association with an outflow is uncertain.

lambda_min_6 = 1525
v_6 = c * (lambda_0 - lambda_min_6) / lambda_0

print(f"J102038: absorption blueshift ≈ {v_6:.0f} km/s")
```

```python
# SDSS J023252.80-001351.1
# Multiple absorption minima; approximate selected minimum.

lambda_min_7 = 1520
v_7 = c * (lambda_0 - lambda_min_7) / lambda_0

print(f"J023252: ≈ {v_7:.0f} km/s")
```

```python
# SDSS J033917.02-051443.7
# Very noisy absorption: this estimate is especially uncertain.

lambda_min_8 = 1465
v_8 = c * (lambda_0 - lambda_min_8) / lambda_0

print(f"J033917: ≈ {v_8:.0f} km/s (very uncertain)")
```

```python
# SDSS J131504.50+500239.5
# Approximate selected minimum within the broad absorption trough.

lambda_min_9 = 1535
v_9 = c * (lambda_0 - lambda_min_9) / lambda_0

print(f"J131504: ≈ {v_9:.0f} km/s")
```

```python
# SDSS J203724.05-002055.1
# Narrow absorption feature; association with an outflow is uncertain.

lambda_min_10 = 1510
v_10 = c * (lambda_0 - lambda_min_10) / lambda_0

print(f"J203724: absorption blueshift ≈ {v_10:.0f} km/s")
```

**Note:** We estimate velocities from the bottoms of the absorption dips. Dips to the left of 1549 Å are blueshifted, meaning the gas moves toward us. Broad dips indicate quasar winds; narrow dips are less certain. If no clear dip is visible, we cannot measure a velocity.

## Conclusion 

The estimated velocities range from about 2,700 to 16,300 km/s. J033917 has the highest estimated velocity, although its noisy spectrum makes this result very uncertain. J131504 has the lowest measured value.

J003832 and J075757 show relatively high velocities, around 10,500 km/s. Identical results for some objects reflect our approximate visual measurements, rather than necessarily identical gas speeds.

For J102038 and J203724, the narrow absorption dips indicate blueshifted gas, but their connection to quasar winds is uncertain. No velocity was measured for J014552 because it lacks a clear absorption dip.

These values describe the selected absorption minima, not the maximum wind speeds.
