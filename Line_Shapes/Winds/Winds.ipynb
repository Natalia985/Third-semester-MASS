We estimate the Hα absorption minimum at each epoch and calculate:

$
v \approx c\,\frac{\lambda_0-\lambda_{\min}}{\lambda_0},
\qquad \lambda_0=6562.8\ \text{Å}.
$

```python
c = 300000       
lambda_0 = 6562.8  
lambda_min = 6300 # approximate visual estimate, Å

v_20 = c * (lambda_0 - lambda_min) / lambda_0

print(f"Day 20: {v_20:.0f} km/s")
```
