# H0 SIS 

```python
import numpy as np
from scipy.integrate import quad

# Integrand 1/E(z) for comoving distance, where E(z) is dimensionless
# Hubble constant at redshift z (eq. 14 in Hogg 2000)
def ez(z, om):
    return 1.0 / np.sqrt(om * np.power(1.0 + z, 3.0) + 1.0 - om)

# Dimensionless comoving distance (eq. 15 in Hogg 2000)
def dc(z, om):
    return quad(ez, 0.0, z, args=(om))[0]

# Constants: Omega_M and 1 Mpc in km
om = 0.27
mpc = 3.08567758e+19

# User inputs
zd = float(input('Redshift of the lens: '))
zs = float(input('Redshift of the source: '))
ra = float(input('Radius of the first image (in arcsec): '))
rb = float(input('Radius of the second image (in arcsec): '))
dt = float(input('Measured time delay (in days): '))

# Convert ra and rb from arcsec to radians
ra = ra * np.pi / 648000.0
rb = rb * np.pi / 648000.0

# Convert dt from days to seconds
dt = dt * 86400.0

# Calculate Hubble constant (in km/s/Mpc)
h0 = dc(zs, om) * dc(zd, om) * (ra * ra - rb * rb) * mpc / (2.0 * dt * (dc(zs, om) - dc(zd, om)))

print('\nHubble constant: H0 = {0:.2f} km/s/Mpc'.format(h0))
```
