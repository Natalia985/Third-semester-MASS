```python
import numpy as np
from scipy.integrate import quad
import matplotlib.pyplot as plt

# Use Computer Modern font
plt.rcParams.update({
    'font.size': 20,
    'font.family': 'serif',
    'font.serif': ['cmr10'],
    'mathtext.fontset': 'cm',
    'axes.formatter.use_mathtext': True
})

# Integrand for dimensionless comoving distance
def ez(z, om):
    return 1.0 / np.sqrt(om * np.power(1.0 + z, 3.0) + 1.0 - om)

# Dimensionless comoving distance (scalar version)
def dc_scalar(z, om):
    return quad(ez, 0.0, z, args=(om))[0]

# Vectorized version of the dimensionless comoving distance
dc = np.vectorize(dc_scalar)

# Dimensionless time-delay distance
def ddt(zd, zs, om):
    return dc(zd, om) * dc(zs, om) / (dc(zs, om) - dc(zd, om))

# Effective lens distances as functions of source redshift
zd = 0.5
zmax = 5.0
zz = np.arange(zd + 0.01, zmax, 0.01)

d1 = ddt(zd, zz, 0.3)
d2 = ddt(zd, zz, 0.0)
d3 = ddt(zd, zz, 1.0)

# Plot the results
plt.plot(zz, d1, 'r-', label=r'$\Omega_M=0.3$')
plt.plot(zz, d2, 'g-', label=r'$\Omega_M=0.0$')
plt.plot(zz, d3, 'b-', label=r'$\Omega_M=1.0$')
plt.gca().set_box_aspect(0.75)
plt.xlim(zd, zmax)
plt.ylim(0.5, 3.0)
plt.xlabel('$z_s$')
plt.ylabel('$d_{\\Delta t}$')
plt.title('$z_d=0.5$')
plt.legend(loc='upper right')

# Save the figure
plt.tight_layout()
plt.savefig('ddt.png', dpi=300, bbox_inches='tight')
plt.show()
```
